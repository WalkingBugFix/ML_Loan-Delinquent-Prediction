```python
x = data.drop(['isDelinquent'], axis=1)
y = data['isDelinquent']
x = add_constant(x)
x = pd.get_dummies(x, drop_first=True).astype(int)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.35, random_state=1, stratify=y)
```

```python
lg = sm.Logit(y_train, x_train).fit()
```

```python
pred_train = np.round(lg.predict(x_train))
```

```python
cm = confusion_matrix(y_train, pred_train)
custom_cm = pd.DataFrame([[cm[1,1],cm[1,0]],
                           [cm[0,1],cm[0,0]]],
                         columns=['Predicted Positive','Predicted Negative'],
                         index=['Actual Positive','Actual Negative'])
custom_cm
```

```python
accuracy_score(y_train, pred_train)
```

### Guassian NB

```python
df.head()
```

```python
df['isDelinquent'].value_counts(1)*100
```

```python
x_nb = df.drop(['isDelinquent'], axis=1)
y_nb = df['isDelinquent']
x_nb = pd.get_dummies(x_nb , drop_first=True)
x_train, x_test, y_train, y_test = train_test_split(x_nb, y_nb, test_size = 0.35, stratify = y, random_state = 1)
```

```python
model = GaussianNB()
model.fit(x_train, y_train.ravel())
```

```python
pred_train = model.predict(x_train)
```

```python
print(classification_report(y_train, pred_train))
```

```python
cm = confusion_matrix(y_train, pred_train)
custom_cm = pd.DataFrame([[cm[1,1],cm[1,0]],
                          [cm[0,1],cm[0,0]]],
                         columns=['Predicted Positive','Predicted Negative'],
                         index=['Actual Positive','Actual Negative'])
custom_cm
```

### Predictions on the test set

```python
pred_test = model.predict(x_test)
```

```python
print(classification_report(y_test, pred_test))
```

```python
cm = confusion_matrix(y_test, pred_test)
custom_cm = pd.DataFrame([[cm[1,1],cm[1,0]],
                          [cm[0,1],cm[0,0]]],
                         columns=['Predicted Positive','Predicted Negative'],
                         index=['Actual Positive','Actual Negative'])
custom_cm
```
